```python
# Import library
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import drive
from google.colab.patches import cv2_imshow

drive.mount('/content/drive')
```

```python
# Membaca gambar
img = cv2.imread('/content/drive/MyDrive/PCVK/ktp_timpa_teks.png')

# Menampilkan gambar menggunakan OpenCV
cv2_imshow(img)

# Menampilkan gambar menggunakan Matplotlib
plt.imshow(img)
plt.title('Gambar dengan Matplotlib')
plt.show()

# Menampilkan ukuran gambar
print("Resolusi gambar:", img.shape)
```

```python
# Konversi BGR menjadi RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Menampilkan gambar RGB
plt.imshow(img_rgb)
plt.title('Citra RGB')
plt.axis('off')
plt.show()

# Konversi menjadi grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Menampilkan grayscale
plt.imshow(img_gray, cmap='gray')
plt.title('Citra Grayscale')
plt.axis('off')
plt.show()
```

```python
# Mengubah ukuran gambar
img_output = cv2.resize(img_rgb, (347, 229))

plt.imshow(img_output)
plt.title('Hasil Resize')
plt.axis('off')
plt.show()

# Melakukan flipping horizontal
img_flip = cv2.flip(img_output, 1)

plt.imshow(img_flip)
plt.title('Hasil Flipping Horizontal')
plt.axis('off')
plt.show()
```

```python
# Membuat canvas hitam
black_img = np.zeros(
    shape=(530, 530, 3),
    dtype=np.int16
)

# Membuat rectangle pertama
cv2.rectangle(
    black_img,
    pt1=(250, 20),
    pt2=(350, 130),
    color=(255, 0, 0),
    thickness=10
)

# Membuat rectangle kedua
cv2.rectangle(
    black_img,
    pt1=(250, 250),
    pt2=(350, 350),
    color=(0, 0, 255),
    thickness=15
)

# Membuat lingkaran
cv2.circle(
    black_img,
    center=(150, 150),
    radius=60,
    color=(0, 255, 0),
    thickness=8
)

# Membuat lingkaran penuh
cv2.circle(
    black_img,
    center=(400, 400),
    radius=50,
    color=(0, 255, 0),
    thickness=-1
)

# Membuat garis diagonal
cv2.line(
    black_img,
    pt1=(0, 0),
    pt2=(530, 530),
    color=(255, 255, 255),
    thickness=5
)

# Menampilkan hasil
plt.imshow(black_img)
plt.title('Bentuk Geometri 2D')
plt.axis('off')
plt.show()
```

```python
# Menentukan font
font = cv2.FONT_HERSHEY_SIMPLEX

# Menambahkan teks
black_img = black_img.astype(np.uint8)
cv2.putText(
    black_img,
    'Laksa',
    (10, 500),
    font,
    2,
    (0, 255, 255),
    5,
    cv2.LINE_AA
)

# Menampilkan hasil
plt.imshow(black_img)
plt.title('Geometri dan Teks')
plt.axis('off')
plt.show()
```

```python
# Membuat canvas hitam kedua
black_img2 = np.zeros(
    shape=(530, 530, 3),
    dtype=np.int32
)

# Membuat titik-titik polygon
vertices = np.array(
    [
        [150, 350],
        [250, 250],
        [450, 350],
        [250, 450]
    ],
    dtype=np.int32
)

# Mengubah bentuk array titik
pts = vertices.reshape((-1, 1, 2))

# Menggambar polyline
cv2.polylines(
    black_img2,
    [pts],
    isClosed=True,
    color=(0, 255, 0),
    thickness=5
)

# Menampilkan hasil
plt.imshow(black_img2)
plt.title('Polyline')
plt.axis('off')
plt.show()
```

# Tugas Penerapan



```python
# Tugas 1
# Kombinasi kanal Merah-Biru (RB)
img_rb = img.copy()
img_rb[:, :, 1] = 0  # channel Green dinolkan

# Kombinasi kanal Hijau-Biru (GB)
img_gb = img.copy()
img_gb[:, :, 2] = 0  # channel Red dinolkan

# Konversi ke RGB agar tampilannya benar di Matplotlib
img_rb_rgb = cv2.cvtColor(img_rb, cv2.COLOR_BGR2RGB)
img_gb_rgb = cv2.cvtColor(img_gb, cv2.COLOR_BGR2RGB)

# Menampilkan keduanya berdampingan
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rb_rgb)
plt.title('Kombinasi Kanal Merah-Biru')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_gb_rgb)
plt.title('Kombinasi Kanal Hijau-Biru')
plt.axis('off')

plt.show()
```

```python
# Tugas 2
# Flipping
flip_vertical = cv2.flip(img, 0)
flip_horizontal = cv2.flip(img, 1)
flip_both = cv2.flip(img, -1)

# Konversi ke RGB agar warna tampil benar di Matplotlib
flip_vertical = cv2.cvtColor(flip_vertical, cv2.COLOR_BGR2RGB)
flip_horizontal = cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB)
flip_both = cv2.cvtColor(flip_both, cv2.COLOR_BGR2RGB)

# Menampilkan ketiga hasil
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertical)
plt.title('Pencerminan Vertikal')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title('Pencerminan Horizontal')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_both)
plt.title('Pencerminan Vertikal + Horizontal')
plt.axis('off')

plt.show()
```

```python
# Studi Kasus 1
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Membaca dua citra dengan ukuran berbeda
img1 = cv2.imread('/content/drive/MyDrive/PCVK/gambar1.jpg')
img2 = cv2.imread('/content/drive/MyDrive/PCVK/gambar2.jpg')


# FOTO 1
h1, w1 = img1.shape[:2]

# Menentukan ukuran bar berdasarkan tinggi gambar
bar_height1 = int(h1 * 0.10)

# Membuat salinan gambar
hasil1 = img1.copy()

# Membuat bar pada bagian bawah gambar
cv2.rectangle(
    hasil1,
    (0, h1 - bar_height1),
    (w1, h1),
    (0, 0, 0),
    -1
)

# Ukuran teks menyesuaikan tinggi gambar
font_scale1 = h1 / 500
thickness1 = max(1, int(h1 / 150))

# Menambahkan nama
cv2.putText(
    hasil1,
    'LAKSAMANA',
    (int(w1 * 0.03), int(h1 - bar_height1 * 0.30)),
    cv2.FONT_HERSHEY_SIMPLEX,
    font_scale1,
    (255, 255, 255),
    thickness1,
    cv2.LINE_AA
)


# FOTO 2
h2, w2 = img2.shape[:2]

# Menentukan ukuran bar berdasarkan tinggi gambar
bar_height2 = int(h2 * 0.10)

# Membuat salinan gambar
hasil2 = img2.copy()

# Membuat bar pada bagian bawah gambar
cv2.rectangle(
    hasil2,
    (0, h2 - bar_height2),
    (w2, h2),
    (0, 0, 0),
    -1
)

# Ukuran teks menyesuaikan tinggi gambar
font_scale2 = h2 / 500
thickness2 = max(1, int(h2 / 150))

# Menambahkan nama
cv2.putText(
    hasil2,
    'LAKSAMANA',
    (int(w2 * 0.03), int(h2 - bar_height2 * 0.30)),
    cv2.FONT_HERSHEY_SIMPLEX,
    font_scale2,
    (255, 255, 255),
    thickness2,
    cv2.LINE_AA
)


# MENAMPILKAN HASIL
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(hasil1, cv2.COLOR_BGR2RGB))
plt.title(f'Foto 1: {w1} x {h1}')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(hasil2, cv2.COLOR_BGR2RGB))
plt.title(f'Foto 2: {w2} x {h2}')
plt.axis('off')

plt.show()
```

```python
# Studi Kasus 2
# Membaca dua gambar
img3 = cv2.imread('/content/drive/MyDrive/PCVK/gambar3.jpg')
img4 = cv2.imread('/content/drive/MyDrive/PCVK/gambar4.webp')


def buat_thumbnail(img, nama_toko):

    # Mendapatkan ukuran gambar asli
    h, w = img.shape[:2]

    # Menentukan ukuran canvas persegi
    size = max(h, w)

    # Membuat canvas putih
    canvas = np.full(
        (size, size, 3),
        255,
        dtype=np.uint8
    )

    # Menghitung ukuran baru dengan mempertahankan rasio
    if w > h:
        new_w = size
        new_h = int(h * size / w)
    else:
        new_h = size
        new_w = int(w * size / h)

    # Resize gambar
    img_resize = cv2.resize(
        img,
        (new_w, new_h)
    )

    # Menghitung posisi agar gambar berada di tengah
    x = int((size - new_w) / 2)
    y = int((size - new_h) / 2)

    # Menempatkan gambar di tengah canvas
    canvas[
        y:y + new_h,
        x:x + new_w
    ] = img_resize

    # Menentukan ukuran tulisan secara proporsional
    font_scale = size / 600
    thickness = max(1, int(size / 150))

    # Menentukan posisi teks
    text_x = int(size * 0.04)
    text_y = int(size * 0.95)

    # Menambahkan nama toko
    cv2.putText(
        canvas,
        nama_toko,
        (text_x, text_y),
        cv2.FONT_HERSHEY_SIMPLEX,
        font_scale,
        (0, 0, 0),
        thickness,
        cv2.LINE_AA
    )

    return canvas


# MEMPROSES KEDUA GAMBAR

hasil3 = buat_thumbnail(
    img3,
    'Toko wowo mart'
)

hasil4 = buat_thumbnail(
    img4,
    'Toko wowo mart'
)

# MENAMPILKAN HASIL
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(hasil3, cv2.COLOR_BGR2RGB))
plt.title('Thumbnail gambar3.jpg')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(hasil4, cv2.COLOR_BGR2RGB))
plt.title('Thumbnail gambar4.webp')
plt.axis('off')

plt.show()


# Menampilkan informasi ukuran
print("Gambar 3:")
print("Ukuran asli     :", img3.shape[1], "x", img3.shape[0])
print("Ukuran thumbnail:", hasil3.shape[1], "x", hasil3.shape[0])

print("\nGambar 4:")
print("Ukuran asli     :", img4.shape[1], "x", img4.shape[0])
print("Ukuran thumbnail:", hasil4.shape[1], "x", hasil4.shape[0])
```
