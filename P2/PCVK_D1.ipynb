```python
# Import library
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import drive
from google.colab.patches import cv2_imshow

drive.mount('/content/drive')
```

```python
# Membaca gambar
img = cv2.imread('/content/drive/MyDrive/PCVK/ktp_timpa_teks.png')

# Menampilkan gambar menggunakan OpenCV
cv2_imshow(img)

# Menampilkan gambar menggunakan Matplotlib
plt.imshow(img)
plt.title('Gambar dengan Matplotlib')
plt.show()

# Menampilkan ukuran gambar
print("Resolusi gambar:", img.shape)
```

```python
# Konversi BGR menjadi RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Menampilkan gambar RGB
plt.imshow(img_rgb)
plt.title('Citra RGB')
plt.axis('off')
plt.show()

# Konversi menjadi grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Menampilkan grayscale
plt.imshow(img_gray, cmap='gray')
plt.title('Citra Grayscale')
plt.axis('off')
plt.show()
```

```python
# Mengubah ukuran gambar
img_output = cv2.resize(img_rgb, (347, 229))

plt.imshow(img_output)
plt.title('Hasil Resize')
plt.axis('off')
plt.show()

# Melakukan flipping horizontal
img_flip = cv2.flip(img_output, 1)

plt.imshow(img_flip)
plt.title('Hasil Flipping Horizontal')
plt.axis('off')
plt.show()
```

```python
# Membuat canvas hitam
black_img = np.zeros(
    shape=(530, 530, 3),
    dtype=np.int16
)

# Membuat rectangle pertama
cv2.rectangle(
    black_img,
    pt1=(250, 20),
    pt2=(350, 130),
    color=(255, 0, 0),
    thickness=10
)

# Membuat rectangle kedua
cv2.rectangle(
    black_img,
    pt1=(250, 250),
    pt2=(350, 350),
    color=(0, 0, 255),
    thickness=15
)

# Membuat lingkaran
cv2.circle(
    black_img,
    center=(150, 150),
    radius=60,
    color=(0, 255, 0),
    thickness=8
)

# Membuat lingkaran penuh
cv2.circle(
    black_img,
    center=(400, 400),
    radius=50,
    color=(0, 255, 0),
    thickness=-1
)

# Membuat garis diagonal
cv2.line(
    black_img,
    pt1=(0, 0),
    pt2=(530, 530),
    color=(255, 255, 255),
    thickness=5
)

# Menampilkan hasil
plt.imshow(black_img)
plt.title('Bentuk Geometri 2D')
plt.axis('off')
plt.show()
```

```python
# Menentukan font
font = cv2.FONT_HERSHEY_SIMPLEX

# Menambahkan teks
black_img = black_img.astype(np.uint8)
cv2.putText(
    black_img,
    'Laksa',
    (10, 500),
    font,
    2,
    (0, 255, 255),
    5,
    cv2.LINE_AA
)

# Menampilkan hasil
plt.imshow(black_img)
plt.title('Geometri dan Teks')
plt.axis('off')
plt.show()
```

```python
# Membuat canvas hitam kedua
black_img2 = np.zeros(
    shape=(530, 530, 3),
    dtype=np.int32
)

# Membuat titik-titik polygon
vertices = np.array(
    [
        [150, 350],
        [250, 250],
        [450, 350],
        [250, 450]
    ],
    dtype=np.int32
)

# Mengubah bentuk array titik
pts = vertices.reshape((-1, 1, 2))

# Menggambar polyline
cv2.polylines(
    black_img2,
    [pts],
    isClosed=True,
    color=(0, 255, 0),
    thickness=5
)

# Menampilkan hasil
plt.imshow(black_img2)
plt.title('Polyline')
plt.axis('off')
plt.show()
```
